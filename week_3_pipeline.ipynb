```python
import pandas as pd
import os

file_path = "/content/learnpath_resources_first20.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 records:")
display(df.head())

print("\nDataset information:")
df.info()
```

```python
# Check missing values
print("Missing values:")
display(df.isnull().sum())

# Check duplicate records
print("\nDuplicate rows:", df.duplicated().sum())

# Check duplicate resource IDs
print("Duplicate resource IDs:", df["resource_id"].duplicated().sum())

# Check resource categories
print("\nResource distribution:")
display(df["primary_domain"].value_counts())
```

```python
# Create a copy of the original dataset
cleaned_df = df.copy()

# Remove leading and trailing spaces from text columns
text_columns = cleaned_df.select_dtypes(include="object").columns

for column in text_columns:
    cleaned_df[column] = cleaned_df[column].str.strip()

# Remove exact duplicate rows
before = len(cleaned_df)
cleaned_df = cleaned_df.drop_duplicates()
after = len(cleaned_df)

print("DATA CLEANING RESULTS")
print("---------------------")
print("Original records:", before)
print("Records after cleaning:", after)
print("Duplicate rows removed:", before - after)

display(cleaned_df.head())
```

```python
duplicate_ids = cleaned_df[
    cleaned_df["resource_id"].duplicated(keep=False)
]

if duplicate_ids.empty:
    print("No duplicate resource IDs found.")
else:
    print("Duplicate resource IDs detected:")
    display(duplicate_ids)
```

```python
print("Missing values after cleaning:")
display(cleaned_df.isnull().sum())
```

```python
# Required columns
required_columns = [
    "resource_id",
    "title",
    "description",
    "primary_domain",
    "resource_type",
    "difficulty_level",
    "provider",
    "resource_url",
    "access_type",
    "verification_status"
]

# Check missing columns
missing_columns = [
    col for col in required_columns
    if col not in cleaned_df.columns
]

# Check duplicate IDs
duplicate_ids = cleaned_df["resource_id"].duplicated().sum()

# Check valid categories
valid_domains = [
    "python_programming",
    "data_analysis",
    "introductory_ml"
]

invalid_domains = cleaned_df[
    ~cleaned_df["primary_domain"].isin(valid_domains)
]

# Check URL format
invalid_urls = cleaned_df[
    ~cleaned_df["resource_url"].str.startswith(
        ("https://", "http://"), na=False
    )
]

# Display validation results
print("DATA VALIDATION REPORT")
print("----------------------")

print("Missing columns:", missing_columns)
print("Duplicate IDs:", duplicate_ids)
print("Invalid domains:", len(invalid_domains))
print("Invalid URLs:", len(invalid_urls))

if not missing_columns and duplicate_ids == 0 and len(invalid_domains) == 0 and len(invalid_urls) == 0:
    print("\nVALIDATION PASSED!")
else:
    print("\nVALIDATION FAILED! Please review the issues.")
```

```python
# Export cleaned dataset
output_path = "/content/prepared_table.csv"

cleaned_df.to_csv(output_path, index=False)

print("Dataset exported successfully!")
print("File:", output_path)
print("Total records:", len(cleaned_df))
```

```python
# Read the exported CSV
prepared_df = pd.read_csv(output_path)

print("EXPORT VERIFICATION")
print("-------------------")

print("Original cleaned dataset:", cleaned_df.shape)
print("Exported dataset:", prepared_df.shape)

# Compare both datasets
if cleaned_df.equals(prepared_df):
    print("SUCCESS: Exported dataset matches cleaned dataset.")
else:
    print("WARNING: Dataset differences detected.")

display(prepared_df.head())
```

```python
import pandas as pd
import hashlib
import os

# Original dataset
source_path = "/content/learnpath_resources_first20.csv"

# Function to calculate SHA-256 hash
def calculate_hash(path):
    sha256 = hashlib.sha256()

    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(4096), b""):
            sha256.update(chunk)

    return sha256.hexdigest()


# Repeatable pipeline function
def run_pipeline(input_path, output_path):

    data = pd.read_csv(input_path)

    # Clean whitespace
    text_columns = data.select_dtypes(include="object").columns

    for column in text_columns:
        data[column] = data[column].str.strip()

    # Remove exact duplicates
    data = data.drop_duplicates()

    # Validate required fields
    required = [
        "resource_id", "title", "description",
        "primary_domain", "resource_url"
    ]

    assert all(col in data.columns for col in required)
    assert data["resource_id"].is_unique
    assert data["resource_url"].notna().all()

    # Export
    data.to_csv(output_path, index=False)

    return data


# Run pipeline twice
run1 = run_pipeline(
    source_path,
    "/content/output_run1.csv"
)

run2 = run_pipeline(
    source_path,
    "/content/output_run2.csv"
)

# Compare outputs
hash1 = calculate_hash("/content/output_run1.csv")
hash2 = calculate_hash("/content/output_run2.csv")

print("PIPELINE REPEATABILITY TEST")
print("---------------------------")

print("Run 1 records:", len(run1))
print("Run 2 records:", len(run2))

print("Run 1 SHA-256:", hash1)
print("Run 2 SHA-256:", hash2)

if hash1 == hash2:
    print("\nSUCCESS: Pipeline is repeatable!")
else:
    print("\nFAILED: Pipeline outputs are different!")
```
